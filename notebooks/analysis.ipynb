```python
df.drop_duplicates(inplace=True) #removed the duplicates rows
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])  #Convert Date Column
```

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Month Name'] = df['Order Date'].dt.month_name()
```

```python
df
```

```python
# total sales
total_sales = df['Sales'].sum()
print(f"{total_sales:.2f}")
```

```python
# top 10 products
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
print(top_products)
```

```python
# Sales by Region
region_sales = df.groupby('Region')['Sales'].sum()
print(region_sales)
```

```python
# Monthly Sales Trend
monthly_sales = df.groupby('Month Name')['Sales'].sum()
print(monthly_sales)
```

```python
# ordered based on months from jan to dec
month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

df['Month Name'] = pd.Categorical(
    df['Month Name'],
    categories=month_order,
    ordered=True
)

monthly_sales = df.groupby('Month Name')['Sales'].sum().sort_index()

print(monthly_sales)
```

```python
# bar chart

import matplotlib.pyplot as plt
import seaborn as sns

top_products.plot(kind='bar', figsize=(12,5))

plt.title("Top 10 Products")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.xticks(rotation=75)
plt.show()
```

```python
# heatmap

pivot = df.pivot_table(
    values='Sales',
    index = 'Region',
    columns = 'Category',
    aggfunc = 'sum'
)

sns.heatmap(pivot, annot=True)
plt.title('Sales Heatmap')
plt.show()
```

```python
# Time Series Plot

monthly = df.groupby('Order Date')['Sales'].sum()
monthly.plot(figsize=(15,5))
plt.title("Sales Trend Over Time")

plt.show()
```

```python
# Customer Purchase Pattern

customer_orders = df.groupby('Customer Name')['Order ID'].count()
customer_orders.sort_values(ascending=False).head(10)
```
